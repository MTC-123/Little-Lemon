This notebook connects to the **LittleLemonDB** MySQL database using the
`mysql-connector-python` client and implements the required stored procedures:

1. `GetMaxQuantity()`
2. `ManageBooking()`
3. `AddBooking()`
4. `UpdateBooking()`
5. `CancelBooking()`


## Task 1: Connect to the database using a Python client


```python
# Install the connector if needed:
# !pip install mysql-connector-python

import mysql.connector as connector

connection = connector.connect(
    user="root",
    password="your_password",  # replace with your MySQL password
    host="localhost",
)
print("Connection established between Python and MySQL:", connection.is_connected())
```

```python
cursor = connection.cursor()
cursor.execute("USE LittleLemonDB")
print("Database in use:", connection.database)
```

## Task 2: Show all tables in the database


```python
cursor.execute("SHOW TABLES")
for table in cursor.fetchall():
    print(table[0])
```

## Task 3: Query with a table JOIN

Return the full name, contact details and order cost for every customer
who placed an order greater than $60 (for a promotional campaign).


```python
join_query = """
SELECT c.FullName,
       c.PhoneNumber,
       c.Email,
       o.TotalCost AS BillAmount
FROM Customers AS c
INNER JOIN Orders AS o ON c.CustomerID = o.CustomerID
WHERE o.TotalCost > 60;
"""
cursor.execute(join_query)
print(cursor.column_names)
for row in cursor.fetchall():
    print(row)
```

## Task 4: `GetMaxQuantity()` stored procedure

Displays the maximum ordered quantity in the Orders table.


```python
cursor.callproc("GetMaxQuantity")
for result in cursor.stored_results():
    print(result.column_names)
    print(result.fetchall())
```

## Task 5: `ManageBooking()` stored procedure

Checks whether a table is already booked on a given date — this is the
procedure that *reacts to changes in the data* by verifying the booking
status before confirming.


```python
cursor.callproc("ManageBooking", ("2022-11-12", 3))
for result in cursor.stored_results():
    print(result.fetchall())
```

## Task 6: `AddBooking()` stored procedure

Adds a new booking to the Bookings table.


```python
cursor.callproc("AddBooking", (9, 3, 4, "2022-12-30"))
for result in cursor.stored_results():
    print(result.fetchall())
connection.commit()
```

## Task 7: `UpdateBooking()` stored procedure

Updates the date of an existing booking.


```python
cursor.callproc("UpdateBooking", (9, "2022-12-17"))
for result in cursor.stored_results():
    print(result.fetchall())
connection.commit()
```

## Task 8: `CancelBooking()` stored procedure

Cancels (deletes) an existing booking.


```python
cursor.callproc("CancelBooking", (9,))
for result in cursor.stored_results():
    print(result.fetchall())
connection.commit()
```

## Task 9: Verify the Bookings table and close the connection


```python
cursor.execute("SELECT * FROM Bookings")
print(cursor.column_names)
for row in cursor.fetchall():
    print(row)
```
